```python
import pandas as pd

# ── Load FP17k-C ─────────────────────────────────────────────────
df = pd.read_csv(
    'gs://fitzpatrick-dataset/metadata/Fitzpatrick_17k_Dataset.csv')

# Drop rows with unknown FST (fst == 0, stored as -1 in fitzpatrick column)
df = df[df['fst'].between(1, 6)].copy()

# Label = FST directly (1-6) — this gives a 6-class model
# NOT binary. AutoML will classify FP1 through FP6 separately.
df['label'] = df['fst'].astype(int)

# Splits are pre-defined by the FP17k-C researchers
# No manual splitting needed — just filter by partition column
train_df = df[df['partition'] == 'train'].copy()
val_df   = df[df['partition'] == 'val'].copy()
test_df  = df[df['partition'] == 'test'].copy()
```
